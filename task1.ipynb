```python
heads = int(input('Enter the number of heads:'))
legs = int(input('Enter the number of legs:'))
for i in range(1,heads,1):
  p = heads-i
  if(2 * i + 4 * p == legs):
    print('dogs ->',i,end=' ')
    print('chicken ->',p)
```
